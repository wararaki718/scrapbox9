```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 20
np.random.seed(42)
```

```python
mean = np.random.normal(loc=50, scale=5, size=N)
variance = np.random.uniform(low=1, high=10, size=N)
std = np.sqrt(variance)
x = np.arange(N)
```

```python
plt.plot(x, mean, marker='o', label='Mean')
plt.fill_between(x, mean-std, mean+std, color='b', alpha=0.2, label='Std Dev')
plt.xlabel('Dataset Id')
plt.ylabel('Value')
plt.title('Line Plot with Mean and Standard Deviation')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
